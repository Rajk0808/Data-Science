## Import the relevant libraries

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

## Load the data

Load data from the csv file: <i> 'Categorical.csv'</i>.

```python
data = pd.read_csv('Categorical.csv')
```

Remove the duplicate index column from the dataset. 

```python
data
```

## Map the data

Use the <i>'continent'</i> category for this analysis.

```python
mapped_data = data.copy()
mapped_data['continent'] = mapped_data['continent'].map({'North America':0, 'Asia':1, 'Africa':2, 'Europe':3, 'South America':4,'Oceania':5, 'Antarctica':6, 'Seven seas (open ocean)':7})
```

## Select the features

```python
x = mapped_data.iloc[:,3:4]
```

## Clustering

Use 4 clusters initially.

```python
kmeans = KMeans(8)
```

## Clustering results

```python
kmeans.fit(x)
```

```python
mapped_data['Cluster'] = kmeans.fit_predict(x)
```

## Plot the data

```python
plt.scatter(mapped_data['Longitude'],mapped_data['Latitude'],c=mapped_data['Cluster'],cmap='rainbow')
```
