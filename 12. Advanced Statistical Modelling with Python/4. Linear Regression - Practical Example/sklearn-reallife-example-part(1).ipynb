```python
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression
import statsmodels.api as sm
import matplotlib.pyplot as plt
sns.set()
```

## loading data


```python
raw_data = pd.read_csv('1.04. Real-life example (1).csv')
raw_data.head()
```

## Data Preprocessing

Exploring the descriptive statistics of the variables


```python
raw_data.describe(include='all')
```

Determining the Variables of interest.

```python
data = raw_data.drop(['Model'],axis=1)
data.describe(include='all')
```

Dealing with missing Values

```python
data.isnull().sum()
```

Rule of thumb: if you are removing <5% of the observation, you are just free to remove all that have MV.
here in these case the total number of values is 4345 and 5% of it is 213 and 172 , 150 both are less then 213, so just drop the rows.

```python
data_no_mv = data.dropna(axis=0)
```

```python
data_no_mv.describe(include='all')
```

## Exploring the Data

```python
sns.distplot(data_no_mv['Price'])
```

Dealing with the outliers

```python
q = data_no_mv['Price'].quantile(0.99)
data_1 = data_no_mv[data_no_mv['Price'] < q]
data_1.describe(include='all')
```

```python
sns.distplot(data_1['Price'])
```

```python
sns.distplot(data_no_mv['Mileage'])
```

```python
q = data_1['Mileage'].quantile(0.99)
data_2 = data_1[data_1['Mileage']<q]
data_2.describe(include='all')
```

```python
sns.distplot(data_2['EngineV'])
```

```python
data_3 = data_2[data_2['EngineV']<6.5]
data_3.describe(include='all')
```

```python
sns.distplot(data_3['EngineV'])
```

```python
sns.distplot(data_3['Year'])
```

```python
q = data_3['Year'].quantile(0.01)
data_4 = data_3[data_3['Year']>q]
data_4.describe(include='all')
```

```python
sns.distplot(data_4['Year'])
```

```python
data_cleaned = data_4.reset_index(drop = True)
data_cleaned.describe(include='all')
```

```python
data_cleaned.to_csv('cleaned_data.csv',index=False)
```
